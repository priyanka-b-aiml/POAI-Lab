```python
import random

actions = ["Explore", "Exploit"]

for i in range(5):
    action = random.choice(actions)
    print("Step", i + 1, ":", action)
```

```python
def value(x):
    return -(x - 5) ** 2 + 25

current = 0

while current < 10:
    next_state = current + 1
    if value(next_state) > value(current):
        current = next_state
    else:
        break

print("Best State:", current)
print("Value:", value(current))
```

```python
import random
import math

def value(x):
    return -(x - 5) ** 2 + 25

current = random.randint(0, 10)
temperature = 10

while temperature > 1:
    next_state = random.randint(0, 10)
    delta = value(next_state) - value(current)

    if delta > 0:
        current = next_state
    else:
        probability = math.exp(delta / temperature)
        if random.random() < probability:
            current = next_state

    temperature -= 1

print("Best State:", current)
print("Value:", value(current))
```

```python
def value(x):
    return -(x - 5) ** 2 + 25

states = [0, 4, 8]

for i in range(3):
    neighbors = []

    for state in states:
        if state > 0:
            neighbors.append(state - 1)
        neighbors.append(state)
        if state < 10:
            neighbors.append(state + 1)

    neighbors = list(set(neighbors))
    neighbors.sort(key=value, reverse=True)
    states = neighbors[:3]

print("Best States:", states)
print("Best Value:", value(states[0]))
```
