```python
# Minimax Search for a simplified Chess position

class ChessGame:
    def __init__(self):
        # Positive values = White pieces
        # Negative values = Black pieces
        self.board = [
            ['r', '.', '.', '.', 'k', '.', '.', 'r'],
            ['p', 'p', 'p', '.', 'p', 'p', 'p', 'p'],
            ['.', '.', '.', '.', '.', '.', '.', '.'],
            ['.', '.', '.', 'p', '.', '.', '.', '.'],
            ['.', '.', '.', 'P', '.', '.', '.', '.'],
            ['.', '.', '.', '.', '.', '.', '.', '.'],
            ['P', 'P', 'P', '.', 'P', 'P', 'P', 'P'],
            ['R', '.', '.', '.', 'K', '.', '.', 'R']
        ]

    def evaluate(self):
        """Evaluate board using piece values."""
        values = {
            'P': 1, 'N': 3, 'B': 3,
            'R': 5, 'Q': 9, 'K': 100,
            'p': -1, 'n': -3, 'b': -3,
            'r': -5, 'q': -9, 'k': -100
        }

        score = 0

        for row in self.board:
            for piece in row:
                score += values.get(piece, 0)

        return score

    def generate_moves(self, player):
        """
        Simplified move generation.
        Returns a list of possible board states.
        This is NOT a complete chess move generator.
        """
        moves = []

        # Demonstration: allow pawns to move one square forward.
        direction = -1 if player == 'white' else 1
        pawn = 'P' if player == 'white' else 'p'

        for r in range(8):
            for c in range(8):
                if self.board[r][c] == pawn:
                    nr = r + direction

                    if 0 <= nr < 8 and self.board[nr][c] == '.':
                        new_board = [row[:] for row in self.board]
                        new_board[nr][c] = pawn
                        new_board[r][c] = '.'

                        new_game = ChessGame()
                        new_game.board = new_board
                        moves.append(new_game)

        return moves


def minimax(game, depth, maximizing_player):
    if depth == 0:
        return game.evaluate()

    player = 'white' if maximizing_player else 'black'
    moves = game.generate_moves(player)

    if not moves:
        return game.evaluate()

    if maximizing_player:
        best_value = float('-inf')

        for move in moves:
            value = minimax(move, depth - 1, False)
            best_value = max(best_value, value)

        return best_value

    else:
        best_value = float('inf')

        for move in moves:
            value = minimax(move, depth - 1, True)
            best_value = min(best_value, value)

        return best_value


game = ChessGame()

print("Initial board:")
for row in game.board:
    print(row)

result = minimax(game, 3, True)

print("\nBest evaluation using Minimax:", result)
```

```python
# Alpha-Beta Pruning for a simplified Chess position

class ChessGame:
    def __init__(self):
        self.board = [
            ['r', '.', '.', '.', 'k', '.', '.', 'r'],
            ['p', 'p', 'p', '.', 'p', 'p', 'p', 'p'],
            ['.', '.', '.', '.', '.', '.', '.', '.'],
            ['.', '.', '.', 'p', '.', '.', '.', '.'],
            ['.', '.', '.', 'P', '.', '.', '.', '.'],
            ['.', '.', '.', '.', '.', '.', '.', '.'],
            ['P', 'P', 'P', '.', 'P', 'P', 'P', 'P'],
            ['R', '.', '.', '.', 'K', '.', '.', 'R']
        ]

    def evaluate(self):
        values = {
            'P': 1, 'N': 3, 'B': 3,
            'R': 5, 'Q': 9, 'K': 100,
            'p': -1, 'n': -3, 'b': -3,
            'r': -5, 'q': -9, 'k': -100
        }

        score = 0

        for row in self.board:
            for piece in row:
                score += values.get(piece, 0)

        return score

    def generate_moves(self, player):
        moves = []

        direction = -1 if player == 'white' else 1
        pawn = 'P' if player == 'white' else 'p'

        for r in range(8):
            for c in range(8):

                if self.board[r][c] == pawn:
                    nr = r + direction

                    if 0 <= nr < 8 and self.board[nr][c] == '.':
                        new_board = [row[:] for row in self.board]

                        new_board[nr][c] = pawn
                        new_board[r][c] = '.'

                        new_game = ChessGame()
                        new_game.board = new_board

                        moves.append(new_game)

        return moves


def alpha_beta(game, depth, alpha, beta, maximizing_player):

    if depth == 0:
        return game.evaluate()

    player = 'white' if maximizing_player else 'black'
    moves = game.generate_moves(player)

    if not moves:
        return game.evaluate()

    if maximizing_player:

        best_value = float('-inf')

        for move in moves:

            value = alpha_beta(
                move,
                depth - 1,
                alpha,
                beta,
                False
            )

            best_value = max(best_value, value)

            alpha = max(alpha, best_value)

            # Beta cutoff
            if beta <= alpha:
                break

        return best_value

    else:

        best_value = float('inf')

        for move in moves:

            value = alpha_beta(
                move,
                depth - 1,
                alpha,
                beta,
                True
            )

            best_value = min(best_value, value)

            beta = min(beta, best_value)

            # Alpha cutoff
            if beta <= alpha:
                break

        return best_value


game = ChessGame()

print("Initial board:")
for row in game.board:
    print(row)

result = alpha_beta(
    game,
    3,
    float('-inf'),
    float('inf'),
    True
)

print("\nBest evaluation using Alpha-Beta:", result)
```

```python
# Minimax Algorithm for Tic-Tac-Toe

import math

board = [' '] * 9


def print_board():
    print()
    print(board[0], '|', board[1], '|', board[2])
    print('--+---+--')
    print(board[3], '|', board[4], '|', board[5])
    print('--+---+--')
    print(board[6], '|', board[7], '|', board[8])
    print()


def check_winner():
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:

        if board[a] == board[b] == board[c] != ' ':
            return board[a]

    if ' ' not in board:
        return 'draw'

    return None


def minimax(is_maximizing):

    result = check_winner()

    # Terminal states
    if result == 'O':
        return 1

    if result == 'X':
        return -1

    if result == 'draw':
        return 0

    # Computer's turn
    if is_maximizing:

        best_score = -math.inf

        for i in range(9):

            if board[i] == ' ':
                board[i] = 'O'

                score = minimax(False)

                board[i] = ' '

                best_score = max(best_score, score)

        return best_score

    # Human's turn
    else:

        best_score = math.inf

        for i in range(9):

            if board[i] == ' ':
                board[i] = 'X'

                score = minimax(True)

                board[i] = ' '

                best_score = min(best_score, score)

        return best_score


def computer_move():

    best_score = -math.inf
    best_move = None

    for i in range(9):

        if board[i] == ' ':

            board[i] = 'O'

            score = minimax(False)

            board[i] = ' '

            if score > best_score:
                best_score = score
                best_move = i

    board[best_move] = 'O'


print("TIC-TAC-TOE")
print("You = X")
print("Computer = O")

while True:

    print_board()

    # Human move
    move = int(input("Enter position (1-9): ")) - 1

    if move < 0 or move > 8 or board[move] != ' ':
        print("Invalid move!")
        continue

    board[move] = 'X'

    result = check_winner()

    if result is not None:
        print_board()

        if result == 'draw':
            print("Game Draw!")

        else:
            print("Winner:", result)

        break

    # Computer move
    computer_move()

    result = check_winner()

    if result is not None:
        print_board()

        if result == 'draw':
            print("Game Draw!")

        else:
            print("Winner:", result)

        break
```
