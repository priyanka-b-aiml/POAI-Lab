```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

visited = set()
queue = deque(['A'])

while queue:
    node = queue.popleft()
    if node not in visited:
        print(node, end=" ")
        visited.add(node)
        queue.extend(graph[node])
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

visited = set()

def dfs(node):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)
        for neighbor in graph[node]:
            dfs(neighbor)

dfs('A')
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def dls(node, limit):
    if limit < 0:
        return
    print(node, end=" ")
    for neighbor in graph[node]:
        dls(neighbor, limit - 1)

dls('A', 2)
```

```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'G'],
    'F': ['C'],
    'G': ['E']
}

start = 'A'
goal = 'G'

front1 = deque([start])
front2 = deque([goal])

visited1 = {start}
visited2 = {goal}

found = False

while front1 and front2 and not found:
    node1 = front1.popleft()
    for neighbor in graph[node1]:
        if neighbor not in visited1:
            visited1.add(neighbor)
            front1.append(neighbor)
        if neighbor in visited2:
            print("Meeting Node:", neighbor)
            found = True
            break

    if found:
        break

    node2 = front2.popleft()
    for neighbor in graph[node2]:
        if neighbor not in visited2:
            visited2.add(neighbor)
            front2.append(neighbor)
        if neighbor in visited1:
            print("Meeting Node:", neighbor)
            found = True
            break
```
