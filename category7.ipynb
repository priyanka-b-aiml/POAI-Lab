```python
def map_coloring(graph, colors, result, node):
    if node == len(graph):
        return True

    region = list(graph.keys())[node]

    for color in colors:
        valid = True

        for neighbor in graph[region]:
            if neighbor in result and result[neighbor] == color:
                valid = False
                break

        if valid:
            result[region] = color

            if map_coloring(graph, colors, result, node + 1):
                return True

            del result[region]

    return False


graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

colors = ['Red', 'Green', 'Blue']
result = {}

if map_coloring(graph, colors, result, 0):
    print("Map Coloring Solution:")
    for region, color in result.items():
        print(region, ":", color)
else:
    print("No solution exists")
```

```python
from itertools import permutations

letters = 'SENDMORY'
digits = '0123456789'

for p in permutations(digits, len(letters)):
    d = dict(zip(letters, p))

    if d['S'] == '0' or d['M'] == '0':
        continue

    SEND = int(d['S'] + d['E'] + d['N'] + d['D'])
    MORE = int(d['M'] + d['O'] + d['R'] + d['E'])
    MONEY = int(d['M'] + d['O'] + d['N'] + d['E'] + d['Y'])

    if SEND + MORE == MONEY:
        print("SEND =", SEND)
        print("MORE =", MORE)
        print("MONEY =", MONEY)
        break
```

```python
import random

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

colors = ['Red', 'Green', 'Blue']
assignment = {region: random.choice(colors) for region in graph}

for _ in range(1000):
    conflicts = []

    for region in graph:
        for neighbor in graph[region]:
            if assignment[region] == assignment[neighbor]:
                conflicts.append(region)
                break

    if not conflicts:
        break

    region = random.choice(conflicts)

    best_color = min(
        colors,
        key=lambda color: sum(
            color == assignment[neighbor]
            for neighbor in graph[region]
        )
    )

    assignment[region] = best_color

conflicts = sum(
    assignment[region] == assignment[neighbor]
    for region in graph
    for neighbor in graph[region]
)

if conflicts == 0:
    print("Solution Found:")
    for region, color in assignment.items():
        print(region, ":", color)
else:
    print("No solution found")
```
