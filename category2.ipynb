```python
class Agent:
    def act(self, observation):
        if observation == "Dirty":
            return "Clean"
        elif observation == "Clean":
            return "No Action"
        else:
            return "Need More Information"

agent = Agent()

print("Fully Observable:")
print(agent.act("Dirty"))

print("Partially Observable:")
print(agent.act(None))
```

```python
known = {
    "A": "Dirty",
    "B": "Clean"
}

room = input("Enter Room: ")

if room in known:
    print("Known Environment")
    print("Status:", known[room])
else:
    print("Unknown Environment")
```

```python
import random

status = "Dirty"

print("Static Environment")
print("Status:", status)

print("Dynamic Environment")
status = random.choice(["Dirty", "Clean"])
print("Status:", status)
```

```python
choice = int(input("Enter Temperature: "))

if choice < 25:
    print("Discrete Action: Heater ON")
else:
    print("Discrete Action: Heater OFF")

print("Continuous Value:", choice)
```

```python
import random

print("Deterministic")
print("Action:", "Clean")

print("Stochastic")
print("Action:", random.choice(["Clean", "Wait"]))
```

```python
rooms = ["Dirty", "Clean", "Dirty"]

print("Episodic")

for room in rooms:
    if room == "Dirty":
        print("Clean")
    else:
        print("No Action")

print("Sequential")

state = "Dirty"

if state == "Dirty":
    print("Cleaning")
    state = "Clean"

print("Current State:", state)
```
