```python


def vacuum_agent(location, status):
    if status == "Dirty":
        return "Clean the room"
    elif location == "A":
        return "Move to room B"
    else:
        return "Move to room A"


location = input("Enter location (A/B): ")
status = input("Enter status (Clean/Dirty): ")

action = vacuum_agent(location, status)

print("Agent Action:", action)
```

```python


class ModelBasedAgent:
    def __init__(self):
        self.model = {"A": "Unknown", "B": "Unknown"}

    def act(self, location, status):
        self.model[location] = status

        if status == "Dirty":
            return "Clean the room"
        elif self.model["A"] == "Clean" and self.model["B"] == "Clean":
            return "All rooms are clean"
        elif location == "A":
            return "Move to room B"
        else:
            return "Move to room A"


agent = ModelBasedAgent()

location = input("Enter location (A/B): ")
status = input("Enter status (Clean/Dirty): ")

print("Agent Action:", agent.act(location, status))
print("Internal Model:", agent.model)
```

```python


def goal_based_agent(current, goal):
    routes = {
        "A": ["B"],
        "B": ["C"],
        "C": ["D"],
        "D": []
    }

    print("Current Location:", current)
    print("Goal:", goal)

    while current != goal:
        if routes[current]:
            next_location = routes[current][0]
            print("Move from", current, "to", next_location)
            current = next_location
        else:
            print("Goal cannot be reached")
            return

    print("Goal Reached!")


start = input("Enter starting location (A/B/C/D): ")
goal = input("Enter goal location (A/B/C/D): ")

goal_based_agent(start, goal)
```

```python


def utility_based_agent():
    options = {
        "Bus": 60,
        "Train": 90,
        "Car": 70,
        "Bike": 50
    }

    best_option = max(options, key=options.get)

    print("Travel Options and Utility:")
    for option, utility in options.items():
        print(option, ":", utility)

    print("\nBest Option:", best_option)
    print("Highest Utility:", options[best_option])


utility_based_agent()
```
