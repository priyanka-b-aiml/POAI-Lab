```python
facts = ["fever", "cough"]

rules = {
    "flu": ["fever", "cough", "body_pain"],
    "cold": ["cough", "sneezing"],
    "infection": ["fever", "body_pain"]
}

facts.append("body_pain")

print("Initial Facts:", facts)

for disease, symptoms in rules.items():
    if all(symptom in facts for symptom in symptoms):
        facts.append(disease)
        print("Forward Chaining:", disease, "detected")

def backward_chaining(goal):
    if goal in facts:
        return True
    if goal in rules:
        return all(backward_chaining(symptom) for symptom in rules[goal])
    return False

goal = "flu"
print("Backward Chaining:", goal, "=", backward_chaining(goal))
```

```python
facts = ["obstacle_detected", "vehicle_moving"]

rules = {
    "apply_brakes": ["obstacle_detected", "vehicle_moving"],
    "stop_vehicle": ["apply_brakes"],
    "safe_to_move": ["road_clear"]
}

print("Initial Facts:", facts)

changed = True
while changed:
    changed = False
    for action, conditions in rules.items():
        if all(condition in facts for condition in conditions) and action not in facts:
            facts.append(action)
            print("Forward Chaining:", action)
            changed = True

def backward_chaining(goal):
    if goal in facts:
        return True
    if goal in rules:
        return all(backward_chaining(condition) for condition in rules[goal])
    return False

goal = "stop_vehicle"
print("Backward Chaining:", goal, "=", backward_chaining(goal))
```
