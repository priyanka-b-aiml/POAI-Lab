```python
import heapq

def greedy_bfs(graph, heuristic, start, goal):
    pq = [(heuristic[start], start)]
    visited = set()
    parent = {start: None}

    while pq:
        _, current = heapq.heappop(pq)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            return path[::-1]

        for neighbor in graph[current]:
            if neighbor not in visited:
                parent[neighbor] = current
                heapq.heappush(pq, (heuristic[neighbor], neighbor))

    return None


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

heuristic = {
    'A': 6,
    'B': 4,
    'C': 5,
    'D': 7,
    'E': 2,
    'F': 3,
    'G': 0
}

start = 'A'
goal = 'G'

path = greedy_bfs(graph, heuristic, start, goal)

print("Greedy Best-First Search Path:")
print(" -> ".join(path))
```

```python
import heapq

def a_star(graph, heuristic, start, goal):
    pq = [(heuristic[start], 0, start)]
    parent = {start: None}
    cost = {start: 0}

    while pq:
        f, g, current = heapq.heappop(pq)

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            return path[::-1], g

        for neighbor, edge_cost in graph[current]:
            new_cost = g + edge_cost

            if neighbor not in cost or new_cost < cost[neighbor]:
                cost[neighbor] = new_cost
                parent[neighbor] = current
                f = new_cost + heuristic[neighbor]
                heapq.heappush(pq, (f, new_cost, neighbor))

    return None, None


graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 2)],
    'D': [('G', 3)],
    'E': [('G', 1)],
    'F': [('G', 5)],
    'G': []
}

heuristic = {
    'A': 6,
    'B': 5,
    'C': 4,
    'D': 3,
    'E': 1,
    'F': 5,
    'G': 0
}

start = 'A'
goal = 'G'

path, cost = a_star(graph, heuristic, start, goal)

print("A* Search Path:")
print(" -> ".join(path))
print("Total Cost:", cost)
```

```python
def ao_star(graph, heuristic, node, solved):
    if node not in graph or not graph[node]:
        solved[node] = True
        return heuristic[node]

    if solved.get(node, False):
        return heuristic[node]

    best_cost = float('inf')
    best_option = None

    for option in graph[node]:
        cost = 0

        for child in option:
            cost += ao_star(graph, heuristic, child, solved)

        if cost < best_cost:
            best_cost = cost
            best_option = option

    heuristic[node] = best_cost
    solved[node] = True

    print(node, "->", best_option, "Cost:", best_cost)

    return best_cost


graph = {
    'A': [['B', 'C'], ['D']],
    'B': [['E'], ['F']],
    'C': [['G']],
    'D': [['H', 'I']],
    'E': [],
    'F': [],
    'G': [],
    'H': [],
    'I': []
}

heuristic = {
    'A': 10,
    'B': 4,
    'C': 2,
    'D': 5,
    'E': 2,
    'F': 3,
    'G': 1,
    'H': 2,
    'I': 2
}

solved = {}

cost = ao_star(graph, heuristic, 'A', solved)

print("AO* Minimum Cost:", cost)
```

```python
import heapq

def sma_star(graph, heuristic, start, goal, memory_limit):
    pq = []
    counter = 0

    heapq.heappush(pq, (heuristic[start], counter, start, [start], 0))

    while pq:
        f, _, node, path, g = heapq.heappop(pq)

        if node == goal:
            return path, g

        for neighbor, cost in graph[node]:
            if neighbor in path:
                continue

            new_g = g + cost
            new_f = new_g + heuristic[neighbor]

            counter += 1
            heapq.heappush(
                pq,
                (new_f, counter, neighbor, path + [neighbor], new_g)
            )

        # Keep only the best nodes within memory limit
        if len(pq) > memory_limit:
            pq = heapq.nsmallest(memory_limit, pq)
            heapq.heapify(pq)

    return None, None


graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 2)],
    'D': [('G', 3)],
    'E': [('G', 1)],
    'F': [('G', 5)],
    'G': []
}

heuristic = {
    'A': 6,
    'B': 5,
    'C': 4,
    'D': 3,
    'E': 1,
    'F': 5,
    'G': 0
}

path, cost = sma_star(
    graph,
    heuristic,
    'A',
    'G',
    memory_limit=3
)

print("Memory-Bounded Heuristic Search Path:")
print(" -> ".join(path))
print("Total Cost:", cost)
```
