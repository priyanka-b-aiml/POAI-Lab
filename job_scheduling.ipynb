```python
import random
import math

jobs = ['A', 'B', 'C', 'D']
profit = {'A': 100, 'B': 80, 'C': 60, 'D': 40}

def score(order):
    return sum(profit[job] for job in order)

current = jobs[:]
random.shuffle(current)
temperature = 100

while temperature > 1:
    new = current[:]

    i, j = random.sample(range(len(jobs)), 2)
    new[i], new[j] = new[j], new[i]

    old_score = score(current)
    new_score = score(new)

    if new_score > old_score:
        current = new
    elif random.random() < math.exp(
        (new_score - old_score) / temperature
    ):
        current = new

    temperature *= 0.9

print("Best job order:", current)
print("Total profit:", score(current))
```
