```python
import matplotlib.pyplot as plt

values = [2, 5, 8, 6, 7, 10, 8, 6, 12, 9, 7]

current = 0
iterations = []
scores = []

print("HILL CLIMBING SEARCH")
print("--------------------")

iteration = 0

while True:
    iterations.append(iteration)
    scores.append(values[current])

    print("Iteration", iteration,
          "Position:", current,
          "Value:", values[current])

    left = current - 1
    right = current + 1
    best = current

    if left >= 0 and values[left] > values[best]:
        best = left

    if right < len(values) and values[right] > values[best]:
        best = right

    if best == current:
        break

    current = best
    iteration += 1

local_maximum = values[current]
global_maximum = max(values)
minimum = min(values)

print("--------------------")
print("Local Maximum :", local_maximum)
print("Global Maximum:", global_maximum)
print("Minimum       :", minimum)
print("Final Position:", current)

plt.figure(figsize=(9, 5))

plt.plot(
    range(len(values)),
    values,
    marker='o',
    linewidth=2,
    label='Search Space'
)

plt.plot(
    iterations,
    scores,
    marker='o',
    linestyle='--',
    linewidth=3,
    label='Hill Climbing Path'
)

global_position = values.index(global_maximum)
minimum_position = values.index(minimum)

plt.scatter(
    current,
    local_maximum,
    color='red',
    s=150,
    label='Local Maximum'
)

plt.scatter(
    global_position,
    global_maximum,
    color='green',
    s=150,
    label='Global Maximum'
)

plt.scatter(
    minimum_position,
    minimum,
    color='blue',
    s=150,
    label='Minimum'
)

plt.xlabel("State / Position")
plt.ylabel("Value")
plt.title("Hill Climbing Search")
plt.legend()
plt.grid(True)
plt.show()
```
